```python
import pandas as pd
```

```python
data={"Obs":[1,2,3,4,5,6],"Rainfall":["High","High","High","Low","Low","Low"],"Soil_Type":["Loamy","Sandy","Clay","Loamy","Sandy","Clay"],"Yield":["Good","Good","Good","Poor","Poor","Good"]}
df=pd.DataFrame(data=data)
```

```python
df
```

```python
print("For Feature X1=Rainfall")
count_high_good=len(df[(df['Rainfall']=='High') & (df['Yield']=='Good')])
count_high_poor=len(df[(df['Rainfall']=='High') & (df['Yield']=='Poor')])
count_low_good=len(df[(df['Rainfall']=='Low') & (df['Yield']=='Good')])
count_low_poor=len(df[(df['Rainfall']=='Low') & (df['Yield']=='Poor')])
print(f"The number of High input with good output are= {count_high_good}")
print(f"The number of High input with poor output are= {count_high_poor}")
print(f"The number of Low input with good output are= {count_low_good}")
print(f"The number of Low input with poor output are= {count_low_poor}")
```

```python
from copy import Error
print("For Error 1")
print("For Feature X1=Rainfall")
if(count_high_good>count_high_poor):
  print("x=High and y=Good is more")
  error=count_high_poor
  print("Error1=",error)
elif(count_high_good<count_high_poor):
  print("x=High and y=Poor is more")
  error=count_high_good
  print("Error1=",error)
else:
  print("x=High and y=Good and y=Poor are equal")
  error=1
  print("Error1=",error)
Error_1=error
```

```python
print("For Error 2")
print("For Feature X1=Rainfall")
if(count_low_good>count_low_poor):
  print("x=Low and y=Good is more")
  error=count_low_poor
  print("Error1=",error)
elif(count_low_good<count_low_poor):
  print("x=low and y=Poor is more")
  error=count_low_good
  print("Error1=",error)
else:
  print("x=low and y=Good and y=Poor are equal")
  error=1
  print("Error1=",error)
Error_2=error
T_Error1=(Error_1+Error_2)/len(df['Obs'])
```

```python
print("Total Error for X1:Rainfall=",T_Error1)
```

```python
print("For Feature X2=Soil_Type")
count_loamy_good=len(df[(df['Soil_Type']=='Loamy') & (df['Yield']=='Good')])
count_loamy_poor=len(df[(df['Soil_Type']=='Loamy') & (df['Yield']=='Poor')])
count_sandy_good=len(df[(df['Soil_Type']=='Sandy') & (df['Yield']=='Good')])
count_sandy_poor=len(df[(df['Soil_Type']=='Sandy') & (df['Yield']=='Poor')])
count_clay_good=len(df[(df['Soil_Type']=='Clay') & (df['Yield']=='Good')])
count_clay_poor=len(df[(df['Soil_Type']=='Clay') & (df['Yield']=='Poor')])
print(f"The number of High input with good output are= {count_loamy_good}")
print(f"The number of High input with poor output are= {count_loamy_poor}")
print(f"The number of Low input with good output are= {count_sandy_good}")
print(f"The number of Low input with poor output are= {count_sandy_poor}")
print(f"The number of Low input with good output are= {count_clay_good}")
print(f"The number of Low input with poor output are= {count_clay_poor}")
```

```python
print("For Error 1")
print("For Feature x2=Soil_Type")
if(count_loamy_good>count_loamy_poor):
  print("x=High and y=Good is more")
  error=count_loamy_poor
  print("Error1=",error)
elif(count_loamy_good<count_loamy_poor):
  print("x=High and y=Poor is more")
  error=count_loamy_good
  print("Error1=",error)
else:
  print("x=Loamy and y=Good and y=Poor are equal")
  error=1
  print("Error1=",error)
Error_1=error
```

```python
print("For Error 2")
print("For Feature X2=Soil_Type")
if(count_sandy_good>count_sandy_poor):
  print("x=Low and y=Good is more")
  error=count_sandy_poor
  print("Error2=",error)
elif(count_sandy_good<count_sandy_poor):
  print("x=low and y=Poor is more")
  error=count_sandy_good
  print("Error2=",error)
else:
  print("x=Sandy and y=Good and y=Poor are equal")
  error=1
  print("Error2=",error)
Error_2=error
```

```python
print("For Error 2")
print("For Feature X1=Soil_Type")
if(count_clay_good>count_clay_poor):
  print("x=Low and y=Good is more")
  error=count_clay_poor
  print("Error3=",error)
elif(count_clay_good<count_clay_poor):
  print("x=low and y=Poor is more")
  error=count_clay_good
  print("Error3=",error)
else:
  print("x=Clay and y=Good and y=Poor are equal")
  error=1
  print("Error3=",error)
Error_3=error
T_Error2=(Error_1+Error_2+Error_3)/len(df['Obs'])
```

```python
print("Total Error for X2:Soil_Type=",T_Error2)
```

```python
if(T_Error1<T_Error2):
  print("X1:Rainfall is better feature")
else:
  print("X2:Soil_Type is better feature")
```

```python
print("Comparision Table")
data={"Features":["Rainfall","Soil_Type"],"Error_Rank":[T_Error1,T_Error2]}
df=pd.DataFrame(data=data)
print(df)
print("Hence we can see from the above that Rainfall is better feature for yield to be predicted")
```
